### FOR ELSE LOOP

```python
# Write a program take a number from user and find the number is present in above lost or not

l = [11,23,45,67,88,65,55,34,75,77]                
count = 0
num = int(input("Enter the number : "))
for i in l:
    count = count + 1
    if i == num:
        print("The number is present")
else:
    print("The number is not present")

print("Number of iteration : ",count)
```

```python
### break statement :
 - to break the loop and stop it where it is
```

```python
# Write a program take a number from user and find he number is present in above lost or not
# Break statement - used to break the loop

l = [11,23,45,67,88,65,55,34,75,77]                
count = 0
num = int(input("Enter the number : "))
for i in l:
    count = count + 1
    if i == num:
        print("The number is present")
    break                                     ## break statement
else:
    print("The number is not present")

print("Number of iteration : ",count)
```

```python
# Given a list of numbers find the frequency of occurence of distinct numbers

l= [11,22,33,11,33,22,22,11,11]
set(l)
count = 0
for i in set(l):
    count = count + 1
    print(i,"occured ",l.count(i))
```

```python
# Given a list of numbers find the frequency of occurence of distinct numbers

l= [11,22,33,11,33,22,22,11,11]
fre = {}

for i in l:
    if i in fre.keys():
        fre[i] += 1
    else:
        fre[i] = 1
fre
fre.items()

for i,j in fre.items():
    print(i,'occured',j,'times')
```

```python
# Given a list of numbers find the frequency of occurence of distinct numbers
# shortened program

l= [11,22,33,11,33,22,22,11,11]
fre = {}

for i in l:
    if i in fre.keys():
        fre[i] += 1
    else:
        fre[i] = 1

for i,j in fre.items():
    print(i,'occured',j,'times')
```

### NESTED LOOPS
