### TRY AND EXCEPT

```python
try:
    x = int(input('Enter the first number : '))
    y = int(input('Enter the second number : '))

    c = x/y
    print(c)
    
except:
    print('Zero is not allowed!')

print('Statement after')
```

```python
try:
    f = open('hello.txt')
    print('Statement after')

except:
    print("File not found!")
```

```python
try:
    x = 10/0
    s = z + 5
    f = open('hello.txt')

except:
    print('Something went wrong!')
```

```python
try:
    x = 10/0
    s = z + 5
    f = open('hello.txt')

except ZeroDivisionError:
    print('Zero is not allowed!')

except NameError:
    print('Variable does not exist!')

except FileNotFoundError:
    print('File does not exist!')

print('\nStatement after')
```

```python
try:
    x = 10/10
    s = 6 + 5
    #f = open('hello.txt')
    a = 5                                # when the error is other then the mentioned one 

except ZeroDivisionError:
    print('Zero is not allowed!')

except NameError:
    print('Variable does not exist!')

except FileNotFoundError:
    print('File does not exist!')

else:
    print('Something went wrong!')            # when the error is other then the mentioned one 
```

```python
try:
    x = 10/0
    s = z + 5
    f = open('hello.txt')
    
except ZeroDivisionError:
    print('Zero is not allowed!')

except NameError:
    print('Variable does not exist!')

except FileNotFoundError:
    print('File does not exist!')

finally:
    print('Finally code is executing..')
```
