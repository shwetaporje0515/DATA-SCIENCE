```python
## HW
## Make a mini calculator and use all operation in that using if elif else

num1 = int(input("Enter the first number  : "))
num2 = int(input("Enter the second number : "))
opr = input("Enter the operator to perform operation : ")

if opr == '+':
    print("Operator is", opr, "so addition of", num1, "and", num2, "is : ", num1 + num2)
elif opr == '-':
    print("Operator is", opr, "so substraction of", num1, "and", num2, "is : ", num1 - num2)
elif opr == '*':
    print("Operator is", opr, "so multiplication of", num1, "and", num2, "is : ", num1 * num2)
elif opr == '/':
    print("Operator is", opr, "so division of", num1, "and", num2, "is : ", num1 - num2)
elif opr == '//':
    print("Operator is", opr, "so floor division of", num1, "and", num2, "is : ", num1 // num2)
elif opr == '%':
    print("Operator is", opr, "so modulus of", num1, "and", num2, "is : ", num1 % num2)
elif opr == '**':
    print("Operator is", opr, "so power of", num1, "and", num2, "is : ", num1 ** num2)
else:
    print("Invalid operator !!!")
```
