## WHAT IS A FUNCTIONS?
- FUNCTION IS A GROUP OF STATEMENTS THAT PERFORM A SPECIFIC TASK.
- A FUNCTION IS A COLLECTION OF STATEMENTS GROUPED TOGETHER THAT PERFORMS AN OPERATION. WHEN YOU CALL THE RANDOM.RANDINT(A,B) FUNCTION FOR EXAMPLE: THE SYSTEM ACTUALLY EXECUTES THE STATEMENTS IN THE FUNCTION AND RETURNS THE RESULT

### ADVANTAGES OF FUNCTIONS:
- MAKES YOUR CODE MORE ORGANIZED AND MANAGEABLE.
- ENSURES REUSABILITY, AVOIDES CODE REDUNDENCY.(REPITITION OF CODE)

```python
## Write a function which takes three numbers argument and prints the largest number

def larg_no(a,b,c):
    if a>b and a>c:
        print(a," is largest")
    elif b>a and b>c:
        print(b," is Largest")
    else:
        print(c," is largest")

num1 = int(input("Enter the no : "))
num2 = int(input("Enter the no : "))
num3 = int(input("Enter the no : "))

larg_no(num1,num2,num3)
```

```python
## Second largest

def seclarge_no(a,b,c):
    if a>b and b>c:
        print(b," is second largest no.")
    elif b>a and a>c:
        print(a," is second Largest no.")
    else:
        print(c," is second largest")

num1 = int(input("Enter the no : "))
num2 = int(input("Enter the no : "))
num3 = int(input("Enter the no : "))

seclarge_no(num1,num2,num3)
```

```python
## Lowest number

def small_no(a,b,c):
    if a<b and a<c:
        print(a," is smallest no")
    elif b<a and b<c:
        print(b," is smallest no")
    else:
        print(c," is smallest no")

num1 = int(input("Enter the no : "))
num2 = int(input("Enter the no : "))
num3 = int(input("Enter the no : "))

small_no(num1,num2,num3)
```

### POSITIONAL ARGUMENT AND KEYWORD ARGUMENT:

```python
def details(name,age,city):
    print("The details are : ")
    print("Name : ", name)
    print("Age  : ", age)
    print("City : ",city)
```

```python
details('shweta','21','Nashik')
```

```python
details('ankur',40,'pune')
```

```python
details(name = 'ankur', age = 40, city = 'pune')            #Keyword argument - order will not change
```

```python
details(name = 'ankur', city = 'pune', age = 40)
```

```python
### Variable argumets(*) and keyword arguments(**) (args and kwargs)
```

```python
def add(num1, num2):
    return num1 + num2
```

```python
add(2,3)
```

```python
add(2,3,4)           # not possible
```

```python
def add(*nums):               # MULTIPLE Variable arguments
    result = 0
    for i in nums:
        result += i

    return result
```

```python
add(4)
```

```python
add(4,5,6)
```

```python
add(10,15,10)
```

```python
add(1,5,4,6,7,8,4,5,1,9)
```

```python
def person(**details):                      ## Multiple keyword arguments
    print("The details are : ")
    for i,j in details.items():
        print(i,":",j)
```

```python
person(name = 'ankur')
```

```python
person(First_name = 'ankur', Last_name = 'varshney', City = 'pune', Age = 40)
```
