```python
class Person2:
    def __init__(self):
        self.name = None              # = '' , = None 
        self.age  = None              # = 0 , = None
        self.city = None

    def get_details(self, name, age, city):
        self.name = name
        self.age = age
        self.city = city

    def show_details(self):
        print("\nThe details are : ")
        print('Name : ', self.name)
        print("Age  : ", self.age)
        print("City : ", self.city)

    def update_city(self, new_city):
        self.city = new_city
        print("City of {} has been updated to {}".format(self.name, self.city))
        
```

```python
p1 = Person2()
```

```python
p1.name
```

```python
p1.get_details('prapti',21,'burhanpur')
p1.name
```

```python
p1.age
```

```python
p1.city
```

```python
p1.show_details()
```

```python
p1.update_city('pune')
```

```python
p1.show_details()
```

## INHERITANCE:
- IT MEANS ONE CLASS(CHILD) CAN USE THE PROPERTIES AND METHOD TO ANOTHER CLASS(PARENT)
- SIMPLY MEANS ACQUIRING PROPERTIES FROM ANOTHER CLASS

```python
class Person:                    # parent class
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Employee(Person):          # Child class
    pass
```

```python
e1 = Employee('Ankur', 40)
e1.name
```

```python
e1.age
```
