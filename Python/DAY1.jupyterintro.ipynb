# END ARGUMENT
- used to specify what appears after print statement


```python
print('my name is shweta', end ='\n')
print('i am a student')

print('my name is shweta', end ='. ')
print('i am a student')

print('my name is shweta ',end ='. ')
print('\ni am a student')
```

# SEP ARGUMENT
- used to specify what appears between the two outputs of same print statement

```python
print("my name is shweta" , "I am a student" , "I am new in pune.")
print('\n')

print("my name is shweta" , "I am a student" , "I am new in pune." , sep = '   ')
print("my name is shweta" , "I am a student" , "I am new in pune." , sep = ' @@@ ')
```

### Swap 2 numbers

```python
#using third variable

a = 2
b = 3

print("number", a)
print("number", b)

c = a
a = b
b = c

print("\n")
print("number", a)
print("number", b)
```

```python
#without using 3 variable - 1 method

a = 5
b = 6
print("a", a)
print("a", b)

a,b = b,a

print("\n")
print("a", a)
print("b", b)
```

```python
#without using 3 variable - 2 method

q = int(input("enter the number: "))
p = int(input("enter the number: "))

print("enter the number before swapping")
print("q", q)
print("p", p)
        
q = (q + p)
print("q", q)

p = q - p
q = p - q

print("q", q)
print("p", p)
```
