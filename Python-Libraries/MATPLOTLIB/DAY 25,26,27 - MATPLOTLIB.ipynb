```python
x = np.arange(30)
y = x + np.random.randn(30)

plt.figure(figsize = (8,4), dpi = 200)
plt.plot(x,y,'c--o')
plt.title('MY VISUALIZATION')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
x = np.arange(30)
y = x + np.random.randn(30)

plt.figure(figsize = (15,5), dpi = 500)
plt.plot(x,y,'c--o')

plt.xticks(range(0,31,1))
plt.yticks(range(0,31,3))

plt.title('MY VISUALIZATION')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

```python
plt.figure(figsize = (15,5), dpi = 500)

plt.plot(x,x**2,'m--o', label ='Square')
plt.plot(x,x**3,'y--*', label ='Cube')

plt.xticks(range(0,31,1))
plt.yticks(range(0,27001,3000))

plt.title('Square vs Cube')
plt.xlabel('Numbers')
plt.ylabel('Square/Cube')

plt.legend()

plt.savefig('Sqaure Vs Cube', dpi = 500)       # To save the file                   
```

## Scatter Plot

```python
x = np.arange(30)
y = x + np.random.randn(30)

plt.figure(figsize = (8,4), dpi = 200)
plt.scatter(x,y)
```

```python
x = np.arange(30)
y = x + np.random.randn(30)

plt.figure(figsize = (8,4), dpi = 200)
plt.scatter(x,y, c = 'm', s =10)     # As default size of bullet is 30
```

## Bar Plot

```python
x = np.arange(10)
y = x*2

plt.figure(dpi = 200)
plt.bar(x,y)
```

```python
x = np.arange(10)
y = x*2

plt.figure(dpi = 200)
plt.bar(x,y, width = 0.5,color = 'm')
```

## Pie Chart

```python
x = np.array([10,20,30,15,40])
plt.pie(x)
```

```python
x = np.array([10,20,30,15,40])

fruits = ['APPLE' ,'GRAPES', 'ORANGES', 'MANGO', 'BANANA']
plt.pie(x, labels = fruits, autopct = '%1.2f%%')
```

```python
x = np.array([10,20,30,15,40])

plt.figure(figsize = (8,3), dpi = 200)
fruits = ['APPLE' ,'GRAPES', 'ORANGES', 'MANGO', 'BANANA']
plt.pie(x, labels = fruits, autopct = '%1.2f%%')
plt.show()              # The extra coding will be hidden
```
